### Tự động làm sạch nhiều file CSV trong thư mục
Đoạn mã này sẽ quét toàn bộ file `.csv` trong thư mục `Data_Warehouse` và áp dụng các bước làm sạch cơ bản.

```python
import glob

# 1. Định nghĩa đường dẫn thư mục chứa 5 file
folder_path = '/content/drive/MyDrive/project/Data_Warehouse/'
all_files = glob.glob(folder_path + "*.csv")

# 2. Hàm làm sạch dữ liệu mẫu
def clean_data(df):
    df = df.dropna(how='all')
    return df

# 3. Vòng lặp xử lý
data_frames = {}
for file in all_files:
    file_name = os.path.basename(file)
    print(f"Đang xử lý: {file_name}...")

    temp_df = pd.read_csv(file)
    cleaned_df = clean_data(temp_df)

    # Lưu vào dictionary để dùng sau (ví dụ: data_frames['sales.csv'])
    data_frames[file_name] = cleaned_df

print("\nHoàn thành làm sạch tất cả các file!")
```

```python
# Data cleaning

if data_frames:
    first_file = list(data_frames.keys())[0]
    print(f"1. Dữ liệu mẫu từ file: {first_file}")
    display(data_frames[first_file].info())
    display(data_frames[first_file].head())

if data_frames:
    first_file = list(data_frames.keys())[1]
    print(f"2. Dữ liệu mẫu từ file: {first_file}")
    display(data_frames[first_file].info())
    display(data_frames[first_file].head())

if data_frames:
    first_file = list(data_frames.keys())[2]
    print(f"3. Dữ liệu mẫu từ file: {first_file}")
    display(data_frames[first_file].info())
    display(data_frames[first_file].head())

if data_frames:
    first_file = list(data_frames.keys())[3]
    print(f"4.. Dữ liệu mẫu từ file: {first_file}")
    display(data_frames[first_file].info())
    display(data_frames[first_file].head())

if data_frames:
    first_file = list(data_frames.keys())[4]
    print(f"5. Dữ liệu mẫu từ file: {first_file}")
    display(data_frames[first_file].info())
    display(data_frames[first_file].head())


```

```python
# Lấy dataframe bằng tên file cụ thể (Key) thay vì dùng chỉ mục số
df = data_frames['products.csv']

# Xóa các dòng có ProductKey là 210 hoặc 211 // fill not Applicable for color
df.drop(df[df['ProductKey'].isin([210, 211])].index, inplace=True)
df["Color"] = df["Color"].fillna("Not Applicable")
# Kiểm tra lại kết quả
display(df.info())
```

```python
# Kiểm tra các cột thực tế có trong bảng geography
print("Các cột trong geography.csv:")
print(data_frames['geography.csv'].columns.tolist())

# Thử lấy các cột đúng của bảng geography (ví dụ: GeographyKey, City)
df_geography = data_frames['geography.csv'][['GeographyKey', 'City', 'CountryName','PostalCode','SalesTerritoryKey']]
display(df_geography.head())
df_territory = data_frames['geography.csv'][['SalesTerritoryKey', 'SalesTerritoryRegion','SalesTerritoryGroup']]
display(df_territory.head())
display(df_territory.shape)
```

```python
# Loại bỏ các dòng trùng lặp trong bảng territory
df_territory_unique = df_territory.drop_duplicates().reset_index(drop=True)

print(f"Số lượng dòng trước khi xóa trùng: {len(df_territory)}")
print(f"Số lượng dòng sau khi xóa trùng: {len(df_territory_unique)}")

# Hiển thị danh sách vùng lãnh thổ duy nhất
display(df_territory_unique)
```

```python
# Đổi tên các cột trong bảng territory
df_territory_unique = df_territory_unique.rename(columns={
    'SalesTerritoryKey': 'sales_teritory_key',
    'SalesTerritoryRegion': 'sales_teritory_region',
    'SalesTerritoryGroup': 'sales_teritory_group'
})

# Ví dụ: Đổi tên cột cho bảng Sales và Products để đồng bộ
data_frames['sales.csv'] = data_frames['sales.csv'].rename(columns={
    'ProductID': 'product_key',
    'OrderDateKey': 'order_date_key',
    'CustomerKey': 'customer_key',
    'SalesTerritoryKey': 'sales_territory_key',
    'OrderQuantity': 'order_quantity',
    'ProductStandardCost': 'product_standard_cost',
    'Revenue': 'sales_amount',
    'OrderDate': 'order_date'
})


data_frames['products.csv'] = data_frames['products.csv'].rename(columns={
    'ProductKey': 'product_key',
    'ProductSubcategoryKey': 'product_subcategory_key',
    'ProductName': 'product_name',
    'Category': 'category',
    'SubCategory': 'sub_category',
    'StandardCost': 'standard_cost',
    'Color': 'color',
    'ModelName': 'model_name',
    'Status':'status'

})


df_geography = df_geography.rename(columns={
    'GeographyKey': 'geography_key',
    'City': 'city',
    'CountryName': 'country_name',
    'PostalCode': 'postal_code',
    'SalesTerritoryKey': 'sales_territory_key'
})

data_frames['customer.csv'] = data_frames['customer.csv'].rename(columns={
    'CustomerKey': 'customer_key',
    'GeographyKey': 'geography_key',
    'CustomerName': 'customer_name',
    'BirthDate': 'birth_date',
    'MaritalStatus':'marital_status',
    'Gender':'gender'
})

data_frames['date.csv'] = data_frames['date.csv'].rename(columns={
    'DateKey': 'date_key',
    'Date': 'date',
    'Day': 'day_name',
    'Month': 'month_name',
    'Year': 'year'
    })

data_frames['sales.csv'].info()

```

```python
order_map = (
    data_frames['sales.csv'][['customer_key', 'order_date']]
    .drop_duplicates()
    .sort_values(['customer_key', 'order_date'])
    .assign(order_key=lambda df: df.index + 1)
)

data_frames['sales.csv'] = data_frames['sales.csv'].merge(
    order_map,
    on=['customer_key', 'order_date'],
    how='left'
)
```

```python
unique_orders = data_frames['sales.csv']['order_key'].nunique()
total_rows = len(data_frames['sales.csv'])

print(f"Tổng số dòng trong bảng Sales: {total_rows}")
print(f"Số lượng đơn hàng (order_key) duy nhất: {unique_orders}")

```

```python

# Xuất các bảng chính trong dictionary data_frames ra CSV
for file_name, df in data_frames.items():
    output_name = f"{file_name}"
    df.to_csv(output_name, index=False)
    print(f"Đã tạo file: {output_name}")

# Xuất thêm các bảng phụ nếu tồn tại
if 'df_territory_unique' in globals():
    df_territory_unique.to_csv('cleaned_territory.csv', index=False)
    print("Đã tạo file: cleaned_territory.csv")

if 'df_geography' in globals():
    df_geography.to_csv('cleaned_geography_main.csv', index=False)
    print("Đã tạo file: cleaned_geography_main.csv")
```
